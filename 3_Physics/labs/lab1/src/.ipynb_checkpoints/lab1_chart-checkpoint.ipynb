```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
# Исходные данные (50 измерений времени, с)
data = [
    9.68, 8.98, 9.83, 9.08, 8.73, 9.70, 10.16, 9.73, 9.76, 10.16,
    10.23, 10.21, 9.68, 9.71, 9.91, 9.06, 9.76, 10.36, 9.91, 10.45,
    9.78, 9.91, 10.10, 9.90, 9.83, 9.50, 9.96, 9.90, 9.80, 9.66,
    10.25, 10.01, 9.46, 9.88, 10.43, 9.96, 9.95, 10.18, 8.73, 8.86,
    9.56, 9.48, 8.76, 9.18, 9.23, 8.88, 8.88, 9.80, 9.48, 8.96
]

# Параметры нормального распределения (вычислены ранее)
mean = 9.667
sigma = 0.476

# Границы интервалов гистограммы (7 интервалов шириной 0.25)
bins = [8.70, 8.95, 9.20, 9.45, 9.70, 9.95, 10.20, 10.45]
```

```python
# Построение гистограммы
plt.figure(figsize=(10, 6))
# density=True делает высоты столбцов равными плотности вероятности
# (ΔN/(N*Δt)), что позволяет сравнивать с теоретической кривой
counts, bins, patches = plt.hist(data, bins=bins, density=True,
                                 alpha=0.6, color='skyblue',
                                 edgecolor='black', label='Экспериментальные данные')

# Теоретическая кривая нормального распределения
x = np.linspace(8.5, 10.7, 200)
y = norm.pdf(x, loc=mean, scale=sigma)
plt.plot(x, y, 'r-', lw=2, label=f'Нормальное распределение\n($\\mu$={mean:.3f} с, $\\sigma$={sigma:.3f} с)')

# Оформление графика
plt.xlabel('Время, с')
plt.ylabel('Плотность вероятности, с$^{-1}$')
plt.title('Гистограмма распределения результатов измерений\nс наложенной кривой Гаусса')
plt.legend()
plt.grid(linestyle='--', alpha=0.7)
plt.tight_layout()

# Сохранение графика в файл
plt.savefig("lab1_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```
