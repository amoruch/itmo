```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def integral_approx(a, b, n, method):
    x = np.linspace(a, b, n+1)
    h = (b - a) / n
    if method == 'left':
        xi = x[:-1]
        approx = h * np.sum(np.cos(2*xi))
    elif method == 'right':
        xi = x[1:]
        approx = h * np.sum(np.cos(2*xi))
    elif method == 'mid':
        xi = (x[:-1] + x[1:]) / 2
        approx = h * np.sum(np.cos(2*xi))
    elif method == 'trap':
        xi = np.cos(2*x)
        approx = h * (xi[0]/2 + np.sum(xi[1:-1]) + xi[-1]/2)
    else:
        raise ValueError("Unknown method")
    return approx, x, xi  # возвращаем узлы и точки оснащения для отрисовки
```

```python
def plot_riemann(a, b, n, method):
    approx, x_nodes, xi = integral_approx(a, b, n, method)
    x_dense = np.linspace(a, b, 500)
    y_dense = np.cos(2*x_dense)

    plt.figure(figsize=(8,5))
    plt.plot(x_dense, y_dense, 'b-', label='f(x)=cos(2x)')

    # Рисуем прямоугольники/трапеции
    h = (b - a) / n
    for i in range(n):
        x_left = x_nodes[i]
        x_right = x_nodes[i+1]
        if method == 'trap':
            # для трапеции рисуем многоугольник
            y_left = np.cos(2*x_left)
            y_right = np.cos(2*x_right)
            plt.fill([x_left, x_right, x_right, x_left],
                     [0, 0, y_right, y_left],
                     edgecolor='black', facecolor='cyan', alpha=0.5)
        else:
            # прямоугольник
            height = np.cos(2*xi[i])
            plt.bar(x_left, height, width=h, align='edge',
                    edgecolor='black', facecolor='cyan', alpha=0.5)

    plt.title(f"Метод: {method}, n={n}, Приближение={approx:.6f}")
    plt.xlabel('x')
    plt.ylabel('y')
    plt.grid(True)
    plt.show()
```

```python
# Пример использования
plot_riemann(0, np.pi/2, 10, 'mid')
```

```python
plot_riemann(0, np.pi/2, 10, 'left')
```

```python
plot_riemann(0, np.pi/2, 100, 'right')
```

```python
plot_riemann(0, np.pi/2, 10, 'trap')
```
